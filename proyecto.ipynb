```python
""" 1. Escribe una función que reciba una cadena de texto como parámetro y devuelva un diccionario con las frecuencias
de cada letra en la cadena. Los espacios no deben ser considerados"""

def contador_letras(cadena):
    """
    Esta función contará cuántas veces se repiten cada una de las letras de la cadena de texto

    Args:
        cadena (string): Cadena de texto para contar sus letras

    Returns:
        Dict: pares clave-valor de letras y el número de veces que se repite

    """
    frecuencias = {}
    for letra in cadena:
        if letra in frecuencias:
            frecuencias[letra] += 1
        elif letra == ' ':
            pass
        else:
            frecuencias[letra] = 1
    return frecuencias

contador_letras('cuenta las letras')
```

```python
""" 3. Escribe una función que tome una lista de palabras y una palabra objetivo como parámetros. 
La función debe devolver una lista con todas las palabras de la lista original 
que contengan la palabra objetivo."""

def lista_palabra_objetivo(lista_palabras, palabra_objetivo):
    """Esta función realiza una lista en la que se incluyen aquellas palabras de la lista
    de palabras que contangan la palabra objetivo

    Args:
        lista_palabra (list): Lista de palabras dada
        palabra_objetivo (string): Palabra objetivo
    
    returns:
        list: Nueva lista que incluye aquellas que contengan la palabra objetivo
    """
    lista_con_palabra_objetivo = []

    for palabra in lista_palabras:
        if palabra_objetivo in palabra:
            lista_con_palabra_objetivo.append(palabra)

    return lista_con_palabra_objetivo

lista_palabra_objetivo(['rascar', 'haber bebido', 'comer', 'lloraba', 'haber comido', 'haber sentido'], 'haber')
```

```python
""" 4. Genera una función que calcule la diferencia entre los valores de dos listas. 
Usa la función map()"""

def diferencia_listas(lista1, lista2):
    """Esta función calcula la diferencia entre los elementos de dos listas

    Args:
        lista1 (list): Lista a la que se le va a restar la siguiente
        lista2 (list): Lista que resta sus valores a los de la anterior
    return:
        list: Una nueva con los resultados de la diferencia
    """
    def resta(elemento1, elemento2):
        return elemento1 - elemento2
    
    resta_map = map(resta, lista1, lista2)

    return list(resta_map)

lista_numeros1 = [1, 23, 41, 10, 100]

lista_numeros2 = [2, 10, 25, 15, 52]

diferencia_listas(lista_numeros1, lista_numeros2)
```

```python
# 6. Escribe una función que calcule el factorial de un número de manera recursiva.

def calculo_factorial(n):
    if n <= 1:
        return 1
    else:
        return n * calculo_factorial(n - 1)

print(calculo_factorial(4))
```

```python
""" 29. Crea una función que convierta una variable en una cadena de texto y enmascare todos los caracteres con el
carácter '#', excepto los últimos cuatro.""" 

def enmascarar_texto(variable):
    """
    La función que nos sirve para enmascarar con una # a excepción de los 4 últimos caracteres

    Args:
        variable (str): La variable a enmascarar

    Returns:
        str: El string ya enmascarado
    """
    texto = str(variable)
    resultado = ""
    for elemento in range(len(texto)):
        if elemento < len(texto) - 4:
            resultado += '#'
        else:
            resultado += texto[elemento]
    return resultado

print(enmascarar_texto('21418491H'))
```

```python
""" 32. Crea una función que tome un nombre completo y una lista de empleados, busque el nombre completo en la lista y
devuelve el puesto del empleado si está en la lista, de lo contrario, devuelve un mensaje indicando que la persona
no trabaja aquí.""" 

def puesto_empleados(nombre_completo, lista_empleados):
    """
    Función que devuelve el puesto del empleado, en base a una lista que le demos nosotros

    Args:
        nombre_completo (str): El nombre del empleado
        lista_empleados (list): La lista de empleados 

    Returns:
        str: Devuelve el nombre y el puesto del empleado en un print
    """
    
    nombre_completo = nombre_completo.upper()

    for empleado in lista_empleados:
        if empleado['nombre'].upper() == nombre_completo:
            return print(f'El puesto del empleado {empleado['nombre']} es {empleado['puesto']}')
        
    return print(f'{nombre_completo},no trabaja aquí')
        
lista_empleados_32 = [{'nombre' : 'Jose', 'puesto' : 'Junior'},
                    {'nombre' : 'Martín', 'puesto' : 'Senior'},
                    {'nombre' : 'Alba','puesto' : 'Project Manager'}]

puesto_empleados('Alba',lista_empleados_32)
puesto_empleados('Luis',lista_empleados_32)
```

```python
""" 34. Crea la clase Arbol , define un árbol genérico con un tronco y ramas como atributos. Los métodos disponibles son:
crecer_tronco , nueva_rama , crecer_ramas , quitar_rama e info_arbol . El objetivo es implementar estos métodos para
manipular la estructura del árbol.
Código a seguir:
1. Inicializar un árbol con un tronco de longitud 1 y una lista vacía de ramas.
2. Implementar el método crecer_tronco para aumentar la longitud del tronco en una unidad.
3. Implementar el método nueva_rama para agregar una nueva rama de longitud 1 a la lista de ramas.
4. Implementar el método crecer_ramas para aumentar en una unidad la longitud de todas las ramas existentes.
5. Implementar el método quitar_rama para eliminar una rama en una posición específica.""" 

class arbol:
    """
        Clase que representa un árbol
    """
    def __init__(self,tronco,ramas):
        """
        Constructor de la clase árbol
        Args:
            tronco (int): Atributo que representa el tronco del arbol (longitud)
            ramas (list): Atributo que representa a las ramas (Número de ramas)
        """
        self.tronco = tronco
        self.ramas = ramas

    def crecer_tronco(self):
        """
        Este método hace crecer la longitud del tronco

        Args:
            metros (int): Metros representa a la unidad que mide la longitud del tronco
        """
        self.tronco += 1

    def nueva_rama(self):
        """
        Este método hace crecer una rama nueva

        Args:
            numero_ramas (list): Determina las ramas que hay en el árbol
        """
        self.ramas.append(1)

    def crecer_ramas(self):
        """
        Este método determina la longitud de todas las ramas del árbol

        Args:
            longitud_ramas (int): La longitud en metros de todas las ramas
        """
        self.ramas = [rama +1 for rama in self.ramas]
        
    def quitar_ramas(self, indice):
        """
        Este método elimina la última rama de la lista
        """
        self.ramas.pop(indice)

    def info_arbol(self):
        """
        Este método muestra la información del arbol
        """
        print(f'La longitud del tronco es: {self.tronco}')
        print(f'El árbol tiene {len(self.ramas)} ramas')
        print(f'Esta es nuestra lista de ramas con sus longitudes {self.ramas}')

"""Caso de uso:
1. Crear un árbol.
2. Hacer crecer el tronco del árbol una unidad.
3. Añadir una nueva rama al árbol.
4. Hacer crecer todas las ramas del árbol una unidad.
5. Añadir dos nuevas ramas al árbol.
6. Retirar la rama situada en la posición 2.
7. Obtener información sobre el árbol.""" 

arbol1 = arbol(1,[])

arbol1.crecer_tronco()
arbol1.nueva_rama()
arbol1.crecer_ramas()
arbol1.nueva_rama()
arbol1.nueva_rama()
arbol1.quitar_ramas(1)
arbol1.info_arbol()
```

```python
""" 36. Crea la clase UsuarioBanco ,representa a un usuario de un banco con su nombre, saldo y si tiene o no cuenta
corriente. Proporciona métodos para realizar operaciones como retirar dinero, transferir dinero desde otro usuario y
agregar dinero al saldo.
Código a seguir:
1. Inicializar un usuario con su nombre, saldo y si tiene o no cuenta corriente mediante True y False .
2. Implementar el método retirar_dinero para retirar dinero del saldo del usuario. Lanzará un error en caso de no
poder hacerse.
3. Implementar el método transferir_dinero para realizar una transferencia desde otro usuario al usuario actual.
Lanzará un error en caso de no poder hacerse.
4. Implementar el método agregar_dinero para agregar dinero al saldo del usuario.""" 

class UsuarioBanco:
    """
    Esta clase representa a un usuario de un banco
    """
    def __init__(self, nombre, saldo, cuenta):
        """
        Aquí definimos los atributos de la clase

        Args:
            nombre (str): El nombre del usuario del banco
            saldo (float): El saldo del que dispone el usuario
            cuenta (bool): Determina si tiene cuenta corriente o no
        """
        self.nombre = nombre
        self.saldo = saldo
        self.cuenta = cuenta

    def retirar_dinero(self, dinero):
        """
        Con este metodo podremos retirar dinero del saldo del usuario

        Args:
            dinero (float): El dinero que el usuario quiere retirar
        """
        if self.saldo <= 0 or dinero > self.saldo:
                raise ValueError(f'{self.nombre} no tienes suficiente dinero en la cuenta')
            
        self.saldo -= dinero
        print(f'Se ha retirado {dinero}€ de la cuenta bancaria {self.nombre}')
            

    def transferir_dinero(self, usuario2, dinero):
        """
        Este método permitirá al usuario hacer transferencia a otro

        Args:
            transferencia (float): La cantidad a transferir
        """
        try:    
            if self.saldo < dinero or self.saldo <= 0:
                raise ValueError(f'{usuario2.nombre} No tiene suficiente dinero en la cuenta') 

            self.saldo -= dinero
            usuario2.saldo += dinero
            print(f'{self.nombre} ha transferido {dinero}€ a la cuenta de {usuario2.nombre}')
        except ValueError:
            print(f'{self.nombre} No tiene suficiente dinero en la cuenta')
            

    def agregar_dinero(self, dinero):
        """
        Con este metodo podremos ingresar dinero a la cuenta del usuario

        Args:
            dinero (float): El dinero a ingresar en la cuenta
        """
        self.saldo += dinero
        print(f'{self.nombre} ha ingresado {dinero}€ en su cuenta')


""" Caso de uso:
1. Crear dos usuarios: "Alicia" con saldo inicial de 100 y "Bob" con saldo inicial de 50, ambos con cuenta corriente.
2. Agregar 20 unidades de saldo de "Bob".
3. Hacer una transferencia de 80 unidades desde "Bob" a "Alicia".
4. Retirar 50 unidades de saldo a "Alicia"."""

alicia = UsuarioBanco('Alicia', 100, True)
bob = UsuarioBanco('Bob', 50, True)

bob.agregar_dinero(20)
bob.transferir_dinero(alicia, 80)
alicia.retirar_dinero(50)
print(f'{alicia.nombre} tiene un saldo de {alicia.saldo}€ en su cuenta')
print(f'{bob.nombre} tiene un saldo de {bob.saldo}€ en su cuenta')
```

```python
""" 37. Crea una función llamada procesar_texto que procesa un texto según la opción especificada: contar_palabras ,
reemplazar_palabras , eliminar_palabra . Estas opciones son otras funciones que tenemos que definir primero y llamar dentro
de la función procesar_texto .
Código a seguir:
1. Crear una función contar_palabras para contar el número de veces que aparece cada palabra en el texto. Tiene
que devolver un diccionario.
2. Crear una función reemplazar_palabras para remplazar una palabra_original del texto por una palabra_nueva . Tiene
que devolver el texto con el remplazo de palabras.
3. Crear una función eliminar_palabra para eliminar una palabra del texto. Tiene que devolver el texto con la palabra
eliminada.
4. Crear la función procesar_texto que tome un texto, una opción(entre "contar", "reemplazar", "eliminar") y un
número de argumentos variable según la opción indicada.
Caso de uso:
Comprueba el funcionamiento completo de la función procesar_texto""" 

def contar_palabras(texto):
    """
    Función que nos permite contar las palabras de un texto

    Args:
        texto (str): El texto que vamos a procesar

    Returns:
        dict: Diccionario de palabras y su contador
    """
    texto = texto.lower().split()
    contador_palabras = {}
    for palabra in texto:
        if palabra in contador_palabras:
            contador_palabras[palabra] += 1
        else:
            contador_palabras[palabra] = 1
    return contador_palabras

def reemplazar_palabras(texto, palabra_reemplazada, palabra_reemplazo):
    """
    La función que nos permite reemplazar palabras del texto

    Args:
        texto (str): El texto que vamos a procesar
        palabra_reemplazada (str): Palabra que se reemplaza
        palabra_reemplazo (str): Palabra nueva

    Returns:
        _type_: _description_
    """
    return texto.replace(palabra_reemplazada,palabra_reemplazo)

def eliminar_palabra(texto, palabra_eliminada):
    """
    La función que nos permitirá eliminar palabras del texto

    Args:
        texto (str): El texto que vamos a procesar
        palabra_eliminada (str): La palabra a eliminar

    Returns:
        str: El texto modificado
    """
    lista_palabras = texto.split()
    texto_eliminado = []
    for palabra in lista_palabras:
        if palabra != palabra_eliminada:
            texto_eliminado.append(palabra)
    return " ".join(texto_eliminado)

def procesar_texto(texto, opcion, *args):
    """
    Con esta función elegimos entre una de las 3 funciones para procesar el texto (contar, reemplazar o eliminar)

    Args:
        texto (str): El texto que vamos a procesar
        opcion (str): La opción a seleccionar para procesar el texto

    Returns:
        dict/str: 
            - En el caso de contar devolvemos el diccionario en el que cuenta cada una de las palabras
            - Para reemplazar y eliminar, devuelve el texto modificado
    """
    opcion = opcion.lower()
    if opcion == 'contar':
        return contar_palabras(texto)
    elif opcion == 'reemplazar' and len(args) == 2:
        return reemplazar_palabras(texto,args[0],args[1])
    elif opcion == 'eliminar':
        return eliminar_palabra(texto,args[0])
    else:
        return print('Selecciona una de la 3 opciones: contar, reemplazar o eliminar')
    
texto_prueba = 'Este texto es un texto de prueba con la prueba podemos contar palabras dentro del texto reemplazarlas y eliminarlas'

print(procesar_texto(texto_prueba,'Contar'))
print(procesar_texto(texto_prueba, 'reemplazar','texto','descripción'))
print(procesar_texto(texto_prueba, 'eliminar', 'prueba'))
```

```python
""" 40. Escribe una función que tome dos parámetros: figura (una cadena que puede ser "rectangulo" , "circulo" o
"triangulo" ) y datos (una tupla con los datos necesarios para calcular el área de la figura).""" 

import math

def area_figura(figura, datos):
    """
    Esta función trata de calcular el area de una figura dada

    Args:
        figura (str): Determina la figura, debe de ser un triangulo, rectangulo o circulo
        datos (tuple): Los datos necesarios para calcular el area en función de la figura

    Raises:
        ValueError: Se ha establecido una figura incorrecta

    Returns:
        float: Devuelve el area de la figura
    """

    try:

        figura = figura.lower()
        if figura == 'rectangulo':
            base, altura = datos
            area = base * altura
        elif figura == 'triangulo':
            base, altura = datos
            area = (base * altura) / 2
        elif figura == 'circulo':
            (radio,) = datos
            area = math.pi * radio**2
        else:
            raise ValueError

        return round(area,2)
    
    except ValueError:
        print('El valor no es válido, debe de ser un rectangulo, triangulo o circulo')

print(area_figura('Triangulo',(3,4)))
print(area_figura('CIRCULO', (3,)))
print(area_figura('rectangulo', (4,4)))
area_figura('cuadrado', (2,2))
```
